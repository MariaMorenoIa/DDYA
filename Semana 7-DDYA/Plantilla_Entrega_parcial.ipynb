*Justificación función T[n]:* 

En el contexto del sistema de inventario de la bodega, si los códigos ingresan con ciertas secuencias (como la inserción de 20, 25, 28 en tus datos de prueba), el Árbol Binario de Búsqueda (BST) se deforma y crea ramas alargadas parecidas a una lista. Esto obliga al sistema a buscar código por código de manera lineal, degradando la eficiencia a $O(n)$. En cambio, el Árbol AVL, al aplicar rotaciones automáticas tras registrar cada producto, redistribuye los códigos uniformemente. Esto asegura que, al buscar cualquier código de producto, se descarte la mitad del árbol en cada decisión, manteniendo un rendimiento óptimo de $O(\log n)$ sin importar el orden de llegada de los datos.   2. Función $T[n]$ para la gestión de productosEl comportamiento matemático de ambas estructuras al insertar o buscar los $n$ productos registrados por el usuario se define así:Para el Árbol AVL (Árbol garantizado y balanceado):
En cada paso por los nodos, el espacio de búsqueda de códigos se divide a la mitad:
   $$T(n) = T(n/2) + c$$
Por el Teorema Maestro, esta recurrencia se resuelve con una complejidad temporal de $O(\log n)$.   Para el BST (Peor caso sin balanceo):
Si los códigos llegan en un orden desfavorable, el tamaño del problema solo se reduce en un nodo a la vez (por ejemplo, descendiendo solo por hijos derechos):
   $$T(n) = T(n-1) + c$$
Al expandir esta ecuación, la complejidad temporal es directamente proporcional a la cantidad de códigos, resultando en $O(n)$.

```python
import matplotlib.pyplot as plt
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izquierda = None
        self.derecha = None
        self.altura = 1

class BST:
    def __init__(self):
        self.raiz = None

    def insertar(self, valor):
        self.raiz = self._insertar(self.raiz, valor)

    def _insertar(self, nodo, valor):
        # Si llegamos a una posicion vacia, creamos el nodo
        if nodo is None:
            return Nodo(valor)

        # Menores a la izquierda
        if valor < nodo.valor:
            nodo.izquierda = self._insertar(nodo.izquierda, valor)

        # Mayores a la derecha
        elif valor > nodo.valor:
            nodo.derecha = self._insertar(nodo.derecha, valor)

        # Los repetidos no se permiten
        else:
            print(f"El codigo {valor} ya existe y no sera insertado.")

        return nodo

class AVL:
    def __init__(self):
        self.raiz = None

    def obtener_altura(self, nodo):
        if nodo is None:
            return 0
        return nodo.altura

    def factor_balance(self, nodo):
        if nodo is None:
            return 0

        return (
            self.obtener_altura(nodo.izquierda)
            - self.obtener_altura(nodo.derecha)
        )


    def actualizar_altura(self, nodo):
        altura_izquierda = self.obtener_altura(nodo.izquierda)
        altura_derecha = self.obtener_altura(nodo.derecha)

        nodo.altura = 1 + max(altura_izquierda, altura_derecha)

    
    def rotacion_derecha(self, nodo):
        nueva_raiz = nodo.izquierda
        subarbol = nueva_raiz.derecha

        nueva_raiz.derecha = nodo
        nodo.izquierda = subarbol

        self.actualizar_altura(nodo)
        self.actualizar_altura(nueva_raiz)

        return nueva_raiz

    def rotacion_izquierda(self, nodo):
        nueva_raiz = nodo.derecha
        subarbol = nueva_raiz.izquierda

        nueva_raiz.izquierda = nodo
        nodo.derecha = subarbol

        self.actualizar_altura(nodo)
        self.actualizar_altura(nueva_raiz)

        return nueva_raiz

   
    def insertar(self, valor):
        self.raiz = self._insertar(self.raiz, valor)

    def _insertar(self, nodo, valor):

        
        if nodo is None:
            return Nodo(valor)

        if valor < nodo.valor:
            nodo.izquierda = self._insertar(nodo.izquierda, valor)

        elif valor > nodo.valor:
            nodo.derecha = self._insertar(nodo.derecha, valor)

        else:
            print(f"El codigo {valor} ya existe y no sera insertado.")
            return nodo

       
        self.actualizar_altura(nodo)

        balance = self.factor_balance(nodo)

        if balance > 1 and valor < nodo.izquierda.valor:
            print(
                f"Rotacion simple a la derecha en el nodo {nodo.valor}"
            )
            return self.rotacion_derecha(nodo)

        if balance < -1 and valor > nodo.derecha.valor:
            print(
                f"Rotacion simple a la izquierda en el nodo {nodo.valor}"
            )
            return self.rotacion_izquierda(nodo)

        if balance > 1 and valor > nodo.izquierda.valor:
            print(
                f"Rotacion doble izquierda-derecha en el nodo {nodo.valor}"
            )

            nodo.izquierda = self.rotacion_izquierda(
                nodo.izquierda
            )

            return self.rotacion_derecha(nodo)

       
        if balance < -1 and valor < nodo.derecha.valor:
            print(
                f"Rotacion doble derecha-izquierda en el nodo {nodo.valor}"
            )

            nodo.derecha = self.rotacion_derecha(
                nodo.derecha
            )

            return self.rotacion_izquierda(nodo)

       
        return nodo


def preorden(nodo, recorrido):
    if nodo is not None:
        recorrido.append(nodo.valor)
        preorden(nodo.izquierda, recorrido)
        preorden(nodo.derecha, recorrido)


def inorden(nodo, recorrido):
    if nodo is not None:
        inorden(nodo.izquierda, recorrido)
        recorrido.append(nodo.valor)
        inorden(nodo.derecha, recorrido)




def postorden(nodo, recorrido):
    if nodo is not None:
        postorden(nodo.izquierda, recorrido)
        postorden(nodo.derecha, recorrido)
        recorrido.append(nodo.valor)



def pedir_cantidad():
    while True:
        try:
            cantidad = int(
                input("Ingrese la cantidad de productos: ")
            )

            if cantidad <= 0:
                print("Error: la cantidad debe ser un entero mayor que 0.")
            else:
                return cantidad

        except ValueError:
            print(
                "Error: debe ingresar un numero entero. "
                "Intente nuevamente."
            )


def pedir_codigo(numero, codigos):
    while True:
        try:
            codigo = int(
                input(f"Ingrese el codigo {numero}: ")
            )

            
            if codigo in codigos:
                print(
                    f"Error: el codigo {codigo} ya fue ingresado. "
                    "Ingrese un codigo diferente."
                )
            else:
                return codigo

        except ValueError:
            print(
                "Error: el codigo debe ser un numero entero. "
                "Intente nuevamente."
            )



def asignar_posiciones(nodo, profundidad, posiciones, contador):
    """
    Asigna una posicion (x, y) a cada nodo.
    El eje X sigue el recorrido inorden.
    """

    if nodo is None:
        return contador

    contador = asignar_posiciones(
        nodo.izquierda,
        profundidad + 1,
        posiciones,
        contador
    )

    posiciones[nodo.valor] = (
        contador,
        -profundidad
    )

    contador += 1

    contador = asignar_posiciones(
        nodo.derecha,
        profundidad + 1,
        posiciones,
        contador
    )

    return contador



def dibujar_conexiones(nodo, posiciones, eje):
    if nodo is None:
        return

    x_actual, y_actual = posiciones[nodo.valor]

   
    if nodo.izquierda is not None:
        x_izq, y_izq = posiciones[nodo.izquierda.valor]

        eje.plot(
            [x_actual, x_izq],
            [y_actual, y_izq],
            linewidth=2
        )

        dibujar_conexiones(
            nodo.izquierda,
            posiciones,
            eje
        )

   
    if nodo.derecha is not None:
        x_der, y_der = posiciones[nodo.derecha.valor]

        eje.plot(
            [x_actual, x_der],
            [y_actual, y_der],
            linewidth=2
        )

        dibujar_conexiones(
            nodo.derecha,
            posiciones,
            eje
        )



def dibujar_nodos(nodo, posiciones, eje):
    if nodo is None:
        return

    x, y = posiciones[nodo.valor]

    eje.scatter(
        x,
        y,
        s=1100,
        zorder=2
    )

    eje.text(
        x,
        y,
        str(nodo.valor),
        ha="center",
        va="center",
        fontsize=11,
        fontweight="bold",
        zorder=3
    )

    dibujar_nodos(
        nodo.izquierda,
        posiciones,
        eje
    )

    dibujar_nodos(
        nodo.derecha,
        posiciones,
        eje
    )



def dibujar_arbol(raiz, titulo):
    if raiz is None:
        print(f"No hay nodos para mostrar en {titulo}.")
        return

    posiciones = {}

    asignar_posiciones(
        raiz,
        0,
        posiciones,
        0
    )

    figura, eje = plt.subplots(
        figsize=(12, 7)
    )

    dibujar_conexiones(
        raiz,
        posiciones,
        eje
    )

    dibujar_nodos(
        raiz,
        posiciones,
        eje
    )

    eje.set_title(
        titulo,
        fontsize=16,
        fontweight="bold"
    )

    eje.axis("off")

    xs = [pos[0] for pos in posiciones.values()]
    ys = [pos[1] for pos in posiciones.values()]

    eje.set_xlim(
        min(xs) - 1,
        max(xs) + 1
    )

    eje.set_ylim(
        min(ys) - 1,
        max(ys) + 1
    )

    plt.tight_layout()
    plt.show()


def mostrar_recorridos(nombre, raiz):
    pre = []
    ino = []
    post = []

    preorden(raiz, pre)
    inorden(raiz, ino)
    postorden(raiz, post)

    print(f"\n===== RECORRIDOS {nombre} =====")
    print("Preorden :", pre)
    print("Inorden  :", ino)
    print("Postorden:", post)



def main():

    print("=" * 60)
    print(" ORGANIZACION DE CODIGOS DE PRODUCTOS")
    print(" BST Y AVL")
    print("=" * 60)

    cantidad = pedir_cantidad()

    codigos = []

    for i in range(1, cantidad + 1):
        codigo = pedir_codigo(i, codigos)
        codigos.append(codigo)

    print("\nLista inicial de codigos:")
    print(codigos)

    bst = BST()

    for codigo in codigos:
        bst.insertar(codigo)

    print("\nBST construido correctamente.")


    avl = AVL()

    print("\nConstruccion del AVL:")

    for codigo in codigos:
        print(f"\nInsertando {codigo}...")
        avl.insertar(codigo)

    print("\nAVL construido y balanceado correctamente.")

    mostrar_recorridos(
        "BST",
        bst.raiz
    )

    mostrar_recorridos(
        "AVL",
        avl.raiz
    )


    print("\nGenerando grafica del BST...")
    dibujar_arbol(
        bst.raiz,
        "Arbol Binario de Busqueda (BST)"
    )

    print("\nGenerando grafica del AVL...")
    dibujar_arbol(
        avl.raiz,
        "Arbol AVL"
    )

   

    print("\n" + "=" * 60)
    print(" COMPARACION FINAL")
    print("=" * 60)

    print("El BST mantiene el orden original de insercion.")
    print("El AVL aplica rotaciones cuando un nodo queda desbalanceado.")
    print("Ambos conservan la propiedad de arbol de busqueda.")
    print(
        "Por esto, el recorrido Inorden produce los codigos "
        "ordenados de menor a mayor."
    )

    print("\nPrograma finalizado correctamente.")


if __name__ == "__main__":
    main()
```

```python


def ejecutar_tests():

    print("\n" + "=" * 60)
    print("TESTS APLICADOS")
    print("=" * 60)

    # Datos de prueba
    codigos = [30, 20, 10, 25, 28, 27, 5, 50]

    # --------------------------------------------------------
    # TEST 1 - CREACION DEL BST
    # --------------------------------------------------------

    bst = BST()

    for codigo in codigos:
        bst.insertar(codigo)

    assert bst.raiz is not None
    assert bst.raiz.valor == 30

    print("TEST 1 OK: La raiz del BST es correcta.")

    # --------------------------------------------------------
    # TEST 2 - ESTRUCTURA BASICA DEL BST
    # --------------------------------------------------------

    assert bst.raiz.izquierda.valor == 20
    assert bst.raiz.derecha.valor == 50
    assert bst.raiz.izquierda.izquierda.valor == 10
    assert bst.raiz.izquierda.derecha.valor == 25

    print("TEST 2 OK: La estructura principal del BST es correcta.")

    # --------------------------------------------------------
    # TEST 3 - PREORDEN DEL BST
    # --------------------------------------------------------

    recorrido = []
    preorden(bst.raiz, recorrido)

    esperado = [30, 20, 10, 5, 25, 28, 27, 50]

    assert recorrido == esperado

    print("TEST 3 OK: El Preorden del BST es correcto.")

    # --------------------------------------------------------
    # TEST 4 - INORDEN DEL BST
    # --------------------------------------------------------

    recorrido = []
    inorden(bst.raiz, recorrido)

    esperado = [5, 10, 20, 25, 27, 28, 30, 50]

    assert recorrido == esperado

    print("TEST 4 OK: El Inorden del BST es correcto.")

    # --------------------------------------------------------
    # TEST 5 - POSTORDEN DEL BST
    # --------------------------------------------------------

    recorrido = []
    postorden(bst.raiz, recorrido)

    esperado = [5, 10, 27, 28, 25, 20, 50, 30]

    assert recorrido == esperado

    print("TEST 5 OK: El Postorden del BST es correcto.")

    # --------------------------------------------------------
    # TEST 6 - INORDEN ORDENADO
    # --------------------------------------------------------

    assert recorrido is not None

    recorrido_inorden = []
    inorden(bst.raiz, recorrido_inorden)

    assert recorrido_inorden == sorted(codigos)

    print("TEST 6 OK: El Inorden del BST esta ordenado.")

    # --------------------------------------------------------
    # TEST 7 - CREACION DEL AVL
    # --------------------------------------------------------

    avl = AVL()

    for codigo in codigos:
        avl.insertar(codigo)

    assert avl.raiz is not None

    print("TEST 7 OK: El AVL fue construido correctamente.")

    # --------------------------------------------------------
    # TEST 8 - RAIZ DEL AVL
    # --------------------------------------------------------

    assert avl.raiz.valor == 25

    print("TEST 8 OK: La raiz del AVL es correcta.")

    # --------------------------------------------------------
    # TEST 9 - PREORDEN DEL AVL
    # --------------------------------------------------------

    recorrido = []
    preorden(avl.raiz, recorrido)

    esperado = [25, 10, 5, 20, 28, 27, 30, 50]

    assert recorrido == esperado

    print("TEST 9 OK: El Preorden del AVL es correcto.")

    # --------------------------------------------------------
    # TEST 10 - INORDEN DEL AVL
    # --------------------------------------------------------

    recorrido = []
    inorden(avl.raiz, recorrido)

    esperado = [5, 10, 20, 25, 27, 28, 30, 50]

    assert recorrido == esperado

    print("TEST 10 OK: El Inorden del AVL es correcto.")

    # --------------------------------------------------------
    # TEST 11 - POSTORDEN DEL AVL
    # --------------------------------------------------------

    recorrido = []
    postorden(avl.raiz, recorrido)

    esperado = [5, 20, 10, 27, 50, 30, 28, 25]

    assert recorrido == esperado

    print("TEST 11 OK: El Postorden del AVL es correcto.")

    # --------------------------------------------------------
    # TEST 12 - FACTOR DE BALANCE DE LA RAIZ
    # --------------------------------------------------------

    balance = avl.factor_balance(avl.raiz)

    assert -1 <= balance <= 1

    print("TEST 12 OK: El factor de balance de la raiz es valido.")

    # --------------------------------------------------------
    # TEST 13 - ALTURA DEL AVL
    # --------------------------------------------------------

    assert avl.obtener_altura(avl.raiz) > 0

    print("TEST 13 OK: La altura del AVL fue calculada.")

    # --------------------------------------------------------
    # TEST 14 - ROTACION SIMPLE A LA DERECHA
    # --------------------------------------------------------

    avl_derecha = AVL()

    avl_derecha.insertar(30)
    avl_derecha.insertar(20)
    avl_derecha.insertar(10)

    assert avl_derecha.raiz.valor == 20
    assert avl_derecha.raiz.izquierda.valor == 10
    assert avl_derecha.raiz.derecha.valor == 30

    print("TEST 14 OK: La rotacion simple a la derecha funciona.")

    # --------------------------------------------------------
    # TEST 15 - ROTACION SIMPLE A LA IZQUIERDA
    # --------------------------------------------------------

    avl_izquierda = AVL()

    avl_izquierda.insertar(10)
    avl_izquierda.insertar(20)
    avl_izquierda.insertar(30)

    assert avl_izquierda.raiz.valor == 20
    assert avl_izquierda.raiz.izquierda.valor == 10
    assert avl_izquierda.raiz.derecha.valor == 30

    print("TEST 15 OK: La rotacion simple a la izquierda funciona.")

    print("\n" + "=" * 60)
    print("TODOS LOS TESTS FUERON SUPERADOS CORRECTAMENTE")
    print("=" * 60)

if __name__ == "__main__":
    ejecutar_tests()
```
