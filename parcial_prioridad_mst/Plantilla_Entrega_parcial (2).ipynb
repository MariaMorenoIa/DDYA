### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


**Justificación**

La operación más costosa del programa es ordenar las tareas por prioridad. Se usa un ordenamiento por comparación (burbuja con bandera): se recorre la lista comparando tareas vecinas y se intercambian si están en desorden. Las demás operaciones (imprimir tareas, recorrer dispositivos) recorren la lista una sola vez, así que son O(n) y no dominan el costo total.
Se toma como operación básica **la comparación entre dos tareas**, y *n* es el número de tareas.


**Función T(n)**

- **Mejor caso** (las tareas ya vienen ordenadas): una sola pasada y ninguna comparación sale en desorden, así que el algoritmo se detiene.

 T(n) = n − 1  →  **O(n)**

- **Peor caso** (las tareas vienen en orden inverso): cada pasada compara e intercambia todo.

 T(n) = (n−1) + (n−2) + … + 1 = n(n−1)/2 comparaciones, más el mismo número de intercambios:

 T(n) = n² − n  →  **O(n²)**
 
**Conclusión:** el programa es lineal en el mejor caso y cuadrático en el peor caso. Con pocas tareas la diferencia no se nota, pero con muchas tareas el peor caso crece mucho más rápido, como se ve en la gráfica.

```python
import matplotlib.pyplot as plt
n = list(range(1, 51))
mejor = [k - 1 for k in n]
peor = [k * (k - 1) for k in n]
plt.figure(figsize=(8, 5))
plt.plot(n, mejor, label="Mejor caso: T(n) = n − 1  →  O(n)", lw=2)
plt.plot(n, peor, label="Peor caso: T(n) = n² − n  →  O(n²)", lw=2)
plt.title("Análisis de complejidad: mejor vs. peor caso")
plt.xlabel("n (número de tareas)")
plt.ylabel("T(n) (operaciones)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
#PUNTO1- Cola de prioridad
tareas = [
    ["T01", "Configurar servidor", 6],
    ["T02", "Reparar sensor", 10],
    ["T03", "Actualizar software", 4],
    ["T04", "Revisar cámara", 8],
    ["T05", "Generar respaldo", 3],
    ["T06", "Configurar alarma", 7]
]

print("Arreglo inicial:")
for tarea in tareas:
    print(tarea)


def max_heapify(heap, n, i):
    mayor = i
    izquierdo = 2 * i + 1
    derecho = 2 * i + 2

    if izquierdo < n:
        if heap[izquierdo][2] > heap[mayor][2]:
            mayor = izquierdo

    if derecho < n:
        if heap[derecho][2] > heap[mayor][2]:
            mayor = derecho

    if mayor != i:
        aux = heap[i]
        heap[i] = heap[mayor]
        heap[mayor] = aux

        max_heapify(heap, n, mayor)

def build_max_heap(heap):
    n = len(heap)

    for i in range(n // 2 - 1, -1, -1):
        max_heapify(heap, n, i)

    return heap


build_max_heap(tareas)

print("Max-Heap resultante:")
for tarea in tareas:
    print(tarea)

#PUNTO2- Procesar tareas
def extract_max(heap):
    if len(heap) == 0:
        return None

    mayor = heap[0]
    ultimo = heap.pop()

    if len(heap) > 0:
        heap[0] = ultimo
        max_heapify(heap, len(heap), 0)

    return mayor

for i in range(2):
    procesada = extract_max(tareas)

    print("Tarea procesada:", procesada)
    print("Heap resultante:")
    for tarea in tareas:
        print(tarea)
    print("----------------")

def insertar(heap, tarea):
    heap.append(tarea)
    i = len(heap) - 1

    while i > 0:
        padre = (i - 1) // 2

        if heap[i][2] > heap[padre][2]:
            aux = heap[i]
            heap[i] = heap[padre]
            heap[padre] = aux
            i = padre
        else:
            break


nueva_tarea = [
    "T07",
    "Falla en sistema contra incendios",
    12
]

insertar(tareas, nueva_tarea)

print("Heap después de insertar T07:")
for tarea in tareas:
    print(tarea)

#PUNTO3 - Dispositivos a conectar 

import matplotlib.pyplot as libgrafica


dispositivos = {
    "A": (2, 2),    # Servidor
    "B": (7, 3),    # Cámara
    "C": (4, 7),    # Sensor
    "D": (10, 8),   # Alarma
    "E": (13, 3),   # Router
    "F": (15, 7)    # Panel
}

libgrafica.figure(figsize=(9, 6))

for nombre, coordenadas in dispositivos.items():
    x = coordenadas[0]
    y = coordenadas[1]
    libgrafica.scatter(x, y, s=100)
    libgrafica.text(x + 0.2, y + 0.2, nombre, fontsize=12)

libgrafica.title("Ubicación de los dispositivos")
libgrafica.xlabel("Coordenada X")
libgrafica.ylabel("Coordenada Y")
libgrafica.grid(True)
libgrafica.show()

import math

aristas = []
ids = list(dispositivos.keys())

for i in range(len(ids)):
    for j in range(i + 1, len(ids)):
        a = ids[i]
        b = ids[j]

        x1 = dispositivos[a][0]
        y1 = dispositivos[a][1]
        x2 = dispositivos[b][0]
        y2 = dispositivos[b][1]

        distancia = math.sqrt(
            (x2 - x1)**2 + (y2 - y1)**2
        )

        aristas.append([distancia, a, b])
        print(a, "-", b, ":", round(distancia, 2))


#PUNTO4 - Árbol de expanción mínima 
def encontrar(padre, x):
    if padre[x] != x:
        padre[x] = encontrar(padre, padre[x])
    return padre[x]


def unir(padre, rango, a, b):
    raiz_a = encontrar(padre, a)
    raiz_b = encontrar(padre, b)

    if raiz_a == raiz_b:
        return False

    if rango[raiz_a] < rango[raiz_b]:
        padre[raiz_a] = raiz_b
    elif rango[raiz_a] > rango[raiz_b]:
        padre[raiz_b] = raiz_a
    else:
        padre[raiz_b] = raiz_a
        rango[raiz_a] += 1

    return True

def kruskal(aristas, dispositivos):

    aristas.sort()

    padre = {}
    rango = {}

    for nodo in dispositivos:
        padre[nodo] = nodo
        rango[nodo] = 0

    mst = []
    total = 0

    for arista in aristas:
        peso = arista[0]
        a = arista[1]
        b = arista[2]

        if unir(padre, rango, a, b):
            mst.append(arista)
            total += peso

        if len(mst) == len(dispositivos) - 1:
            break

    return mst, total


mst, total = kruskal(aristas, dispositivos)

print("CONEXIONES DEL MST:")
for peso, a, b in mst:
    print(a, "-", b, ":", round(peso, 2))

print("Número de conexiones:", len(mst))
print("Longitud total:", round(total, 2))


#PUNTO5 - Visualización final
libgrafica.figure(figsize=(10, 7))

# Dibujar los seis dispositivos
for nombre, coordenadas in dispositivos.items():
    x = coordenadas[0]
    y = coordenadas[1]
    libgrafica.scatter(x, y, s=150, color="blue")
    libgrafica.text(x + 0.2, y + 0.2, nombre, fontsize=12)

# Dibujar únicamente las aristas del MST
for peso, a, b in mst:
    x1 = dispositivos[a][0]
    y1 = dispositivos[a][1]
    x2 = dispositivos[b][0]
    y2 = dispositivos[b][1]

    libgrafica.plot([x1, x2], [y1, y2], color="green")

    mitad_x = (x1 + x2) / 2
    mitad_y = (y1 + y2) / 2

    libgrafica.text(
        mitad_x,
        mitad_y,
        str(round(peso, 2)),
        color="red",
        fontsize=10
    )

libgrafica.title("Árbol de Expansión Mínima - Kruskal")
libgrafica.xlabel("Coordenada X")
libgrafica.ylabel("Coordenada Y")
libgrafica.grid(True)

libgrafica.figtext(
    0.5,
    0.02,
    "Longitud total del cable: " + str(round(total, 2)),
    ha="center",
    fontsize=12
)

libgrafica.show()
```
