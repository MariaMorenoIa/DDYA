## 1) Descripción del problema **[1 pts]**


Un equipo de electrónica está diseñando un prototipo sobre una protoboard. Sobre ella se han ubicado diferentes componentes que deben quedar conectados entre sí.

El problema es que se desea utilizar la menor cantidad posible de cable.

Para resolverlo, cada componente será representado como un vértice, cada posible conexión mediante cable será una arista y la longitud del cable será el peso de la arista.

El objetivo será construir un Árbol de Expansión Mínima (MST) que conecte todos los componentes utilizando la menor longitud total de cable.No se evaluará el funcionamiento electrónico del circuito. La protoboard se utilizará como representación espacial del problema.

Se tienen siete componentes ubicados en una protoboard de 30 columnas (1 a 30) y 10 filas (A a J). Cada componente se representa mediante un vértice de un grafo y cada posible cable entre dos componentes se representa mediante una arista.

El peso de cada arista corresponde a la distancia Manhattan entre las posiciones de los componentes:

distancia = |columna1 - columna2| + |fila1 - fila2|

Para convertir las filas en números se utiliza A=1, B=2, C=3, D=4, E=5, F=6, G=7, H=8, I=9 y J=10.

El programa calcula automáticamente las 21 conexiones posibles entre los siete componentes. Después aplica el algoritmo de Kruskal para seleccionar las conexiones de menor peso sin formar ciclos. Como existen siete vértices, el árbol final debe contener exactamente seis aristas.



En el punto 6 se cambia el buzzer de H6 a J20 y se calcula nuevamente el árbol de expansión mínima.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
import matplotlib.pyplot as plt
from itertools import combinations

componentes_iniciales = {
    "F": {"nombre": "Fuente de alimentación", "posicion": "A3"},
    "L": {"nombre": "LED", "posicion": "C8"},
    "S": {"nombre": "Sensor", "posicion": "B16"},
    "P": {"nombre": "Pulsador", "posicion": "D24"},
    "B": {"nombre": "Buzzer", "posicion": "H6"},
    "R": {"nombre": "Potenciómetro", "posicion": "I15"},
    "M": {"nombre": "Microcontrolador", "posicion": "G25"}
}

filas = "ABCDEFGHIJ"

def convertir_posicion(posicion):

    if not isinstance(posicion, str):
        raise ValueError("La posición debe ser texto.")

    posicion = posicion.upper().strip()

    if len(posicion) < 2:
        raise ValueError("Formato incorrecto. Ejemplo: A3.")

    fila = posicion[0]

    try:
        columna = int(posicion[1:])
    except ValueError:
        raise ValueError("La columna debe ser un número.")

    if fila not in filas:
        raise ValueError("La fila debe estar entre A y J.")

    if columna < 1 or columna > 30:
        raise ValueError("La columna debe estar entre 1 y 30.")

    numero_fila = filas.index(fila) + 1

    return columna, numero_fila


def distancia(posicion1, posicion2):
    x1, y1 = convertir_posicion(posicion1)
    x2, y2 = convertir_posicion(posicion2)

    return abs(x1 - x2) + abs(y1 - y2)

def crear_aristas(componentes):
    aristas = []

    identificadores = list(componentes.keys())

    for comp1, comp2 in combinations(identificadores, 2):
        pos1 = componentes[comp1]["posicion"]
        pos2 = componentes[comp2]["posicion"]

        peso = distancia(pos1, pos2)
        aristas.append((peso, comp1, comp2))

    return aristas

def mostrar_aristas(aristas):
    print("\nTODAS LAS CONEXIONES POSIBLES")
    print("--------------------------------")

    for peso, comp1, comp2 in aristas:
        print(comp1, "------", peso, "------", comp2)

    print("\nCantidad de conexiones:", len(aristas))


def buscar(padre, componente):
    
    if padre[componente] != componente:
        padre[componente] = buscar(padre, padre[componente])

    return padre[componente]


def unir(padre, raiz1, raiz2):
    padre[raiz2] = raiz1


def kruskal(componentes, aristas):
    padre = {}

    for componente in componentes:
        padre[componente] = componente

   
    aristas_ordenadas = sorted(aristas, key=lambda arista: arista[0])

    mst = []
    longitud_total = 0

    for peso, comp1, comp2 in aristas_ordenadas:
        raiz1 = buscar(padre, comp1)
        raiz2 = buscar(padre, comp2)

       
        if raiz1 != raiz2:
            unir(padre, raiz1, raiz2)
            mst.append((peso, comp1, comp2))
            longitud_total += peso

        
        if len(mst) == len(componentes) - 1:
            break

    return mst, longitud_total


def mostrar_mst(mst, longitud_total):
    
    print("\nARBOL DE EXPANSION MINIMA (MST)")
    print("--------------------------------")

    for numero, (peso, comp1, comp2) in enumerate(mst, start=1):
        print("Cable", numero, ":", comp1, "------", peso, "------", comp2)

    print("\nCantidad de cables seleccionados:", len(mst))
    print("Longitud total del cable:", longitud_total, "unidades")



def dibujar_protoboard(componentes, aristas=None, titulo="Protoboard"):
   
    plt.figure(figsize=(15, 6))

   
    for columna in range(1, 31):
        for numero_fila in range(1, 11):
            plt.scatter(columna, numero_fila, color="lightgray", s=12)

    
    if aristas is not None:
        for peso, comp1, comp2 in aristas:
            x1, y1 = convertir_posicion(componentes[comp1]["posicion"])
            x2, y2 = convertir_posicion(componentes[comp2]["posicion"])

            
            plt.plot(
                [x1, x2, x2],
                [y1, y1, y2],
                linewidth=1.5
            )
            mitad_x = (x1 + x2) / 2
            mitad_y = (y1 + y2) / 2
            plt.text(
                mitad_x,
                mitad_y,
                str(peso),
                fontsize=9,
                backgroundcolor="white"
            )
           
    
    for identificador, datos in componentes.items():
        x, y = convertir_posicion(datos["posicion"])
        plt.scatter(x, y, s=100, edgecolors="black", zorder=3)
        plt.text(x, y - 0.35, identificador, ha="center",
                 fontsize=10, fontweight="bold")
        plt.text(x, y + 0.35, datos["posicion"], ha="center", fontsize=8)

    plt.xticks(range(1, 31))
    plt.yticks(range(1, 11), list(filas))
    plt.xlim(0.5, 30.5)
    plt.ylim(10.5, 0.5)
    plt.xlabel("Columnas")
    plt.ylabel("Filas")
    plt.title(titulo)
    plt.grid(True, linestyle="--", alpha=0.4)
    plt.tight_layout()
    plt.show()


def cambiar_buzzer(componentes):

    posicion_anterior = componentes["B"]["posicion"]

    componentes["B"]["posicion"] = "J20"

    nuevas_aristas = crear_aristas(componentes)
    nuevo_mst, nueva_longitud = kruskal(componentes, nuevas_aristas)

    return posicion_anterior, nuevas_aristas, nuevo_mst, nueva_longitud



def ejecutar_pruebas():

    componentes = componentes_iniciales.copy()

    componentes = {
        clave: {"nombre": valor["nombre"], "posicion": valor["posicion"]}
        for clave, valor in componentes_iniciales.items()
    }

    aristas = crear_aristas(componentes)
    mst, total = kruskal(componentes, aristas)

    assert len(componentes) == 7
    assert len(aristas) == 21
    assert distancia("A3", "C8") == 7
    assert len(mst) == len(componentes) - 1
    assert total == sum(arista[0] for arista in mst)

    
    conexiones = set()
    for peso, comp1, comp2 in mst:
        conexiones.add(comp1)
        conexiones.add(comp2)
    assert conexiones == set(componentes.keys())

    
    componentes["B"]["posicion"] = "J20"
    nuevas_aristas = crear_aristas(componentes)
    nuevo_mst, nuevo_total = kruskal(componentes, nuevas_aristas)

    assert len(nuevas_aristas) == 21
    assert len(nuevo_mst) == 6
    assert componentes["B"]["posicion"] == "J20"

    print("\nTodas las pruebas se ejecutaron correctamente.")


def menu():
    
    componentes = {
        clave: {"nombre": valor["nombre"], "posicion": valor["posicion"]}
        for clave, valor in componentes_iniciales.items()
    }

    while True:
        print("\n==============================================")
        print(" LABORATORIO: PROTOBOARD Y ARBOL DE KRUSKAL")
        print("==============================================")
        print("1. Visualizar protoboard inicial")
        print("2. Mostrar todas las posibles conexiones")
        print("3. Ejecutar Kruskal y mostrar el MST")
        print("4. Comparar grafo completo y conexiones mínimas")
        print("5. Cambiar buzzer de H6 a J20 y recalcular")
        print("6. Ejecutar pruebas del programa")
        print("0. Salir")

        opcion = input("\nSeleccione una opción: ")

        if opcion == "1":
            dibujar_protoboard(
                componentes,
                titulo="Distribución actual de la protoboard"
            )

        elif opcion == "2":
            aristas = crear_aristas(componentes)
            mostrar_aristas(aristas)
            dibujar_protoboard(
                componentes,
                aristas,
                "Todas las posibles conexiones"
            )

        elif opcion == "3":
            aristas = crear_aristas(componentes)
            mst, total = kruskal(componentes, aristas)
            mostrar_mst(mst, total)
            dibujar_protoboard(
                componentes,
                mst,
                "Conexiones seleccionadas por Kruskal"
            )

        elif opcion == "4":
            aristas = crear_aristas(componentes)
            mst, total = kruskal(componentes, aristas)

            print("\nCOMPARACIÓN VISUAL")
            print("El grafo completo contiene", len(aristas), "conexiones.")
            print("El MST contiene", len(mst), "conexiones.")
            mostrar_mst(mst, total)

            dibujar_protoboard(componentes, titulo="1. Protoboard actual")
            dibujar_protoboard(
                componentes, aristas, "2. Todas las conexiones posibles"
            )
            dibujar_protoboard(
                componentes, mst, "3. Conexiones mínimas (MST)"
            )

        elif opcion == "5":
            posicion_anterior, nuevas_aristas, nuevo_mst, nueva_longitud = (
                cambiar_buzzer(componentes)
            )

            print("\nCAMBIO DE POSICIÓN DEL BUZZER")
            print("Posición original:", posicion_anterior)
            print("Nueva posición:", componentes["B"]["posicion"])

            aristas_originales = crear_aristas(componentes_iniciales)
            mst_original, total_original = kruskal(
                componentes_iniciales, aristas_originales
            )

            print("\nMST ORIGINAL")
            mostrar_mst(mst_original, total_original)

            print("\nMST DESPUÉS DEL CAMBIO")
            mostrar_mst(nuevo_mst, nueva_longitud)

            diferencia = nueva_longitud - total_original
            if diferencia > 0:
                print("La cantidad total de cable aumentó en", diferencia,
                      "unidades.")
            elif diferencia < 0:
                print("La cantidad total de cable disminuyó en",
                      abs(diferencia), "unidades.")
            else:
                print("La cantidad total de cable no cambió.")

            dibujar_protoboard(
                componentes,
                nuevo_mst,
                "MST después de mover el buzzer a J20"
            )

        elif opcion == "6":
            ejecutar_pruebas()

        elif opcion == "0":
            print("Programa finalizado.")
            break

        else:
            print("Opción no válida. Seleccione una opción del menú.")



menu()

```
