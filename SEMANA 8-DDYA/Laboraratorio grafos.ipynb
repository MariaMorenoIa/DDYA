Ejercicio: Recorrido de una red de distribución
Una empresa de mensajería tiene varios centros de distribución conectados mediante rutas terrestres.

Cada centro será representado mediante un vértice y cada carretera que conecta dos centros será representada mediante una arista.

La empresa desea desarrollar un programa que permita recorrer la red utilizando los algoritmos:

BFS (Breadth-First Search).

DFS (Depth-First Search).

Red de distribución
La empresa tiene los siguientes centros:

A - Centro principal B - Centro Norte C - Centro Sur D - Centro Occidente E - Centro Oriente F - Centro Industrial G - Centro Aeropuerto
Las conexiones entre ellos son:

A - B A - C B - D B - E C - F E - G F - G
Gráficamente, la red puede representarse así:

            A           /   \          B     C         / \     \        D   E     F             \   /               G
Punto 1. Representar el grafo
Construya el grafo utilizando una lista de adyacencia.

El programa deberá mostrar las conexiones de cada centro.

Punto 2. Recorrido BFS
El administrador necesita identificar primero los centros que se encuentran más cerca del centro principal en cantidad de conexiones.

Implemente el algoritmo BFS comenzando desde el centro A.

Para ello deberá utilizar una cola.

El recorrido esperado es:

A -> B -> C -> D -> E -> F -> G

Nivel 0: A Nivel 1: B, C Nivel 2: D, E, F Nivel 3: G
Punto 3. Recorrido DFS

Para ello quiere avanzar por una ruta hasta donde sea posible antes de regresar y explorar otra conexión.

Implemente el algoritmo DFS comenzando desde A.

Puede implementarlo mediante recursividad o utilizando una pila.

Un posible recorrido sería:

A -> B -> D -> E -> G -> F -> C
El resultado puede variar dependiendo del orden en que se almacenen los vecinos.

No existe una ruta hasta el centro solicitado.

La empresa está ampliando su red de distribución, por lo que el programa deberá permitir agregar nuevos centros y nuevas conexiones.

El usuario deberá indicar primero el nombre del nuevo centro.

Ejemplo:

Ingrese el nuevo centro: H

## 6) Código funcional **[15 pts]**

```python

import matplotlib.pyplot as plt


grafo = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "G"],
    "F": ["C", "G"],
    "G": ["E", "F"]
}



def pedir_entero(mensaje):
    while True:
        try:
            numero = int(input(mensaje))
            return numero

        except ValueError:
            print("Debe ingresar un número entero.")



def pedir_centro(grafo, mensaje):

    centro = input(mensaje).upper()

    while centro not in grafo:
        print("El centro no existe.")
        centro = input(mensaje).upper()

    return centro



def mostrar_grafo(grafo):

    print("\nLISTA DE ADYACENCIA")

    for centro in grafo:
        print(centro, "->", grafo[centro])



def mostrar_grafico(grafo):

    posiciones = {
        "A": (0, 3),
        "B": (-2, 2),
        "C": (2, 2),
        "D": (-3, 1),
        "E": (-1, 1),
        "F": (2, 1),
        "G": (0, 0)
    }


    plt.figure(figsize=(7,5))


    conexiones = []


    for nodo in grafo:

        for vecino in grafo[nodo]:

            if (vecino, nodo) not in conexiones:

                x1, y1 = posiciones[nodo]
                x2, y2 = posiciones[vecino]

                plt.plot(
                    [x1, x2],
                    [y1, y2]
                )

                conexiones.append((nodo, vecino))



    for nodo in grafo:

        if nodo in posiciones:

            x, y = posiciones[nodo]

            plt.scatter(x, y, s=1200)

            plt.text(
                x,
                y,
                nodo,
                fontsize=14,
                ha="center",
                va="center"
            )


    plt.title("Red de distribución")

    plt.axis("off")

    plt.show()



def bfs_niveles(grafo, inicio):

    visitados = []
    cola = []
    niveles = {}

    cola.append(inicio)
    visitados.append(inicio)

    niveles[inicio] = 0


    while len(cola) > 0:

        actual = cola.pop(0)


        for vecino in grafo[actual]:

            if vecino not in visitados:

                visitados.append(vecino)

                cola.append(vecino)

                niveles[vecino] = niveles[actual] + 1


    return visitados, niveles



def bfs(grafo, inicio):

    recorrido, niveles = bfs_niveles(grafo, inicio)

    return recorrido



def dfs(grafo, nodo, visitados):

    visitados.append(nodo)


    for vecino in grafo[nodo]:

        if vecino not in visitados:

            dfs(grafo, vecino, visitados)



def buscar_bfs(grafo, inicio, destino):

    cola = []
    visitados = []
    padres = {}


    cola.append(inicio)

    visitados.append(inicio)

    padres[inicio] = None



    while len(cola) > 0:

        actual = cola.pop(0)


        if actual == destino:

            ruta = []


            while actual != None:

                ruta.append(actual)

                actual = padres[actual]


            ruta.reverse()

            return ruta



        for vecino in grafo[actual]:

            if vecino not in visitados:

                visitados.append(vecino)

                cola.append(vecino)

                padres[vecino] = actual



    return None



def agregar_centro(grafo):

    nuevo = input("Ingrese el nuevo centro: ").upper()


    while nuevo == "":

        print("El nombre no puede estar vacío.")

        nuevo = input("Ingrese el nuevo centro: ").upper()



    if nuevo in grafo:

        print("El centro ya existe.")

        return



    grafo[nuevo] = []



    while True:

        conexion = input("Conectar con (FIN para terminar): ").upper()


        if conexion == "FIN":

            break



        if conexion in grafo:

            grafo[nuevo].append(conexion)

            grafo[conexion].append(nuevo)

        else:

            print("Ese centro no existe.")



    print("Centro agregado correctamente.")



def menu():

    print("\n------ MENU ------")
    print("1. Mostrar grafo")
    print("2. Recorrido BFS")
    print("3. Recorrido DFS")
    print("4. Buscar centro con BFS")
    print("5. Agregar centro")
    print("6. Salir")



def main():

    opcion = 0


    while opcion != 6:


        menu()


        opcion = pedir_entero("Opción: ")



        if opcion == 1:

            mostrar_grafo(grafo)

            mostrar_grafico(grafo)



        elif opcion == 2:

            inicio = pedir_centro(grafo, "Centro inicial: ")

            recorrido, niveles = bfs_niveles(grafo, inicio)


            print("\nRecorrido BFS:")

            print(recorrido)


            print("\nNiveles:")

            for centro in niveles:

                print("Nivel", niveles[centro], ":", centro)



        elif opcion == 3:

            inicio = pedir_centro(grafo, "Centro inicial: ")

            recorrido = []

            dfs(grafo, inicio, recorrido)


            print("\nRecorrido DFS:")

            print(recorrido)



        elif opcion == 4:

            inicio = pedir_centro(grafo, "Centro inicial: ")

            destino = pedir_centro(grafo, "Centro que desea buscar: ")


            ruta = buscar_bfs(grafo, inicio, destino)


            if ruta:

                print("Centro encontrado")

                print("Ruta:", ruta)

                print("Cantidad de conexiones:", len(ruta)-1)

            else:

                print("No existe una ruta hasta el centro solicitado.")



        elif opcion == 5:

            agregar_centro(grafo)

            mostrar_grafo(grafo)

            mostrar_grafico(grafo)



        elif opcion == 6:

            print("Finalizando programa...")


        else:

            print("Opción inválida")



main()
```
