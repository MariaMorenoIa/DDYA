## 1) Descripción del problema **[1 pts]**


**Ejercicio: Sistema de atención de una tienda**

Una tienda desea desarrollar un pequeño sistema para controlar la atención de clientes, registrar las operaciones realizadas y administrar su catálogo de productos.


Para resolver el problema se deberán utilizar tres estructuras de datos diferentes:


Cola: para administrar los clientes que esperan ser atendidos.

Pila: para guardar el historial de las últimas operaciones realizadas.

Lista doblemente enlazada: para administrar y recorrer el catálogo de productos.

**Punto 1. Cola de clientes**

Los clientes son atendidos en el mismo orden en que llegan a la tienda.


El usuario deberá ingresar inicialmente una cantidad n de clientes.


Ejemplo:


¿Cuántos clientes desea registrar?: 4 Ingrese el nombre del cliente 1: Ana Ingrese el nombre del cliente 2: Carlos Ingrese el nombre del cliente 3: Laura Ingrese el nombre del cliente 4: Mateo
La cola inicialmente sería:


Ana -> Carlos -> Laura -> Mateo

El programa deberá permitir:


Agregar un nuevo cliente a la cola.

Atender al siguiente cliente.

Mostrar los clientes que siguen esperando.

Por ejemplo:


Atendiendo a: Ana Clientes esperando: Carlos -> Laura -> Mateo

Cuando llegue un nuevo cliente:


Ingrese el nuevo cliente: Sofía

La cola deberá quedar:


Carlos -> Laura -> Mateo -> Sofía

**Punto 2. Pila de operaciones**

Cada vez que se atienda un cliente, su nombre deberá guardarse en una pila de clientes atendidos.


Por ejemplo, después de atender:


Ana Carlos Laura

La pila sería conceptualmente:


| Laura  | <- Tope | Carlos | | Ana    | ---------
El programa deberá permitir:


Agregar a la pila cada cliente que sea atendido.

Mostrar el último cliente atendido.

Deshacer la última atención.

Si se selecciona la opción:


Deshacer última atención

y Laura fue la última cliente atendida, deberá retirarse de la pila.


La pila quedaría:


| Carlos | <- Tope | Ana    | ---------
Además, Laura deberá regresar al inicio de la cola de clientes pendientes.


**Punto 3. Lista doblemente enlazada de productos**

La tienda también necesita mantener un catálogo de productos.


Cada nodo deberá almacenar:


Código Nombre Precio

Por ejemplo:


101 - Arroz - $4.500 102 - Leche - $3.800 103 - Pan - $2.500 104 - Café - $12.000
Estos productos deberán almacenarse utilizando una lista doblemente enlazada.


Cada nodo tendrá:


anterior <- Producto -> siguiente
Ejemplo:


NULL <- Arroz <-> Leche <-> Pan <-> Café -> NULL
El programa deberá permitir:


Agregar un producto al final de la lista.

Mostrar todos los productos desde el primero hasta el último.

Mostrar los productos desde el último hasta el primero.

Buscar un producto utilizando su código.

Eliminar un producto utilizando su código.

Por ejemplo:


Ingrese el código que desea eliminar: 103
Después de eliminar el producto Pan, la lista deberá quedar:


NULL <- Arroz <-> Leche <-> Café -> NULL

*Justificación:* El sistema utiliza tres estructuras de datos lineales:
Cola de clientes (FIFO)
Pila de clientes atendidos (LIFO)
Lista doblemente enlazada de productos
La complejidad se analiza teniendo en cuenta el número de elementos almacenados en cada estructura, representado por n.
Se considera que:
La cola tiene referencias al inicio y al final.
La pila tiene una referencia al tope.
La lista doblemente enlazada tiene referencias al primer y último nodo.
Por lo tanto, las operaciones de inserción y eliminación que utilizan directamente estos extremos pueden realizarse en tiempo constante O(1). En cambio, las operaciones que necesitan recorrer los elementos pueden llegar a requerir O(n)..

*Función T[n]:* 


*Mejor y peor caso con gráfica:* 
En términos generales, las estructuras presentan dos comportamientos principales:
**Mejor caso — O(1)**
Ocurre cuando la operación puede realizarse directamente sin recorrer la estructura.
Ejemplos:
Agregar cliente a la cola.
Atender el primer cliente.
Consultar el último cliente atendido.
Deshacer la última atención.
Agregar producto al final.
Buscar un producto que se encuentra en el primer nodo.
El número de operaciones permanece prácticamente constante aunque aumente n.

**Peor caso — O(n)**
Ocurre cuando es necesario recorrer una cantidad proporcional a los elementos almacenados.
Ejemplos:
Mostrar todos los clientes.
Mostrar todo el historial.
Mostrar todos los productos.
Buscar un producto que está al final.
Buscar un producto que no existe.
Eliminar un producto que se encuentra al final.
En estos casos, al aumentar n, aumenta proporcionalmente el número de operaciones.

```python
def pedir_entero(mensaje, minimo=None):
    """
    Solicita un numero entero y valida que sea correcto.
    """
    while True:
        try:
            valor = int(input(mensaje))
            if minimo is not None and valor < minimo:
                print("ERROR: el valor debe ser mayor o igual a", minimo)
            else:
                return valor
        except ValueError:
            print("ERROR: debe ingresar un numero entero.")
            print("Intente nuevamente.")
def pedir_decimal(mensaje, minimo=None):
    """
    Solicita un numero decimal y valida que sea correcto.
    """
    while True:
        try:
            valor = float(input(mensaje))
            if minimo is not None and valor < minimo:
                print("ERROR: el valor debe ser mayor o igual a", minimo)
            else:
                return valor
        except ValueError:
            print("ERROR: debe ingresar un numero.")
            print("Intente nuevamente.")
def pedir_texto(mensaje):
    """
    Solicita un texto y evita que quede vacio.
    """
    while True:
        valor = input(mensaje).strip()

        if valor == "":
            print("ERROR: este campo no puede estar vacio.")
        else:
            return valor
class NodoCliente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.siguiente = None
class ColaClientes:
    def __init__(self):
        self.frente = None
        self.final = None
    def esta_vacia(self):
        return self.frente is None
    def agregar(self, nombre):
        nuevo = NodoCliente(nombre)
        if self.esta_vacia():
            self.frente = nuevo
            self.final = nuevo
        else:
            self.final.siguiente = nuevo
            self.final = nuevo
    def insertar_al_frente(self, nombre):
        nuevo = NodoCliente(nombre)
        if self.esta_vacia():
            self.frente = nuevo
            self.final = nuevo
        else:
            nuevo.siguiente = self.frente
            self.frente = nuevo
    def atender(self):
        if self.esta_vacia():
            return None
        cliente = self.frente
        self.frente = self.frente.siguiente
        if self.frente is None:
            self.final = None
        cliente.siguiente = None
        return cliente.nombre
    def mostrar(self):
        if self.esta_vacia():
            print("\nCOLA VACIA")
            print("No hay clientes en espera.")
            return
        actual = self.frente
        print("\n====================================")
        print("       CLIENTES EN ESPERA")
        print("====================================")
        print("Orden FIFO:")
        print()
        posicion = 1
        while actual is not None:
            print(str(posicion) + ". " + actual.nombre)
            actual = actual.siguiente
            posicion += 1
        print("====================================")
class NodoPila:
    def __init__(self, nombre):
        self.nombre = nombre
        self.siguiente = None
class PilaAtendidos:
    def __init__(self):
        self.tope = None
    def esta_vacia(self):
        return self.tope is None
    def apilar(self, nombre):
        nuevo = NodoPila(nombre)
        nuevo.siguiente = self.tope
        self.tope = nuevo
    def desapilar(self):
        if self.esta_vacia():
            return None
        cliente = self.tope
        self.tope = self.tope.siguiente
        cliente.siguiente = None
        return cliente.nombre
    def ultimo_atendido(self):
        if self.esta_vacia():
            return None
        return self.tope.nombre
    def mostrar(self):
        if self.esta_vacia():
            print("\nHISTORIAL VACIO")
            print("No hay clientes atendidos.")
            return
        actual = self.tope
        print("\n====================================")
        print("       HISTORIAL DE ATENCION")
        print("====================================")
        print("Orden LIFO:")
        print()
        posicion = 1
        while actual is not None:
            print(str(posicion) + ". " + actual.nombre)
            actual = actual.siguiente
            posicion += 1
        print("====================================")
class NodoProducto:
    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.anterior = None
        self.siguiente = None
class ListaDobleProductos:
    def __init__(self):
        self.primero = None
        self.ultimo = None
    def esta_vacia(self):
        return self.primero is None
    def buscar(self, codigo):
        actual = self.primero
        while actual is not None:
            if actual.codigo == codigo:
                return actual
            actual = actual.siguiente
        return None
    def agregar(self, codigo, nombre, precio):
        nuevo = NodoProducto(codigo, nombre, precio)
        if self.esta_vacia():
            self.primero = nuevo
            self.ultimo = nuevo
        else:
            nuevo.anterior = self.ultimo
            self.ultimo.siguiente = nuevo
            self.ultimo = nuevo
    def mostrar_inicio_fin(self):
        if self.esta_vacia():
            print("\nLISTA VACIA")
            print("No hay productos registrados.")
            return
        actual = self.primero
        print("\n====================================")
        print("       CATALOGO DE PRODUCTOS")
        print("       INICIO -> FIN")
        print("====================================")
        while actual is not None:
            print("Codigo :", actual.codigo)
            print("Nombre :", actual.nombre)
            print("Precio : $", format(actual.precio, ".2f"))
            print("------------------------------------")
            actual = actual.siguiente
    def mostrar_fin_inicio(self):
        if self.esta_vacia():
            print("\nLISTA VACIA")
            print("No hay productos registrados.")
            return
        actual = self.ultimo
        print("\n====================================")
        print("       CATALOGO DE PRODUCTOS")
        print("       FIN -> INICIO")
        print("====================================")
        while actual is not None:
            print("Codigo :", actual.codigo)
            print("Nombre :", actual.nombre)
            print("Precio : $", format(actual.precio, ".2f"))
            print("------------------------------------")
            actual = actual.anterior
    def eliminar(self, codigo):
        actual = self.buscar(codigo)
        if actual is None:
            return False
        if actual.anterior is None:
            self.primero = actual.siguiente
        else:
            actual.anterior.siguiente = actual.siguiente
        if actual.siguiente is None:
            self.ultimo = actual.anterior
        else:
            actual.siguiente.anterior = actual.anterior
        actual.anterior = None
        actual.siguiente = None
        return True
def registrar_clientes_iniciales(cola):
    print("\n====================================")
    print("    REGISTRAR CLIENTES INICIALES")
    print("====================================")
    cantidad = pedir_entero(
        "Ingrese la cantidad de clientes: ",
        0
    )
    if cantidad == 0:
        print("No se registraron clientes.")
        return
    for i in range(cantidad):
        nombre = pedir_texto(
            "Ingrese el nombre del cliente "
            + str(i + 1)
            + ": "
        )
        cola.agregar(nombre)
    print("\nClientes registrados correctamente.")
def agregar_nuevo_cliente(cola):
    print("\n====================================")
    print("       AGREGAR NUEVO CLIENTE")
    print("====================================")
    nombre = pedir_texto(
        "Ingrese el nombre del cliente: "
    )
    cola.agregar(nombre)
    print("\nCliente agregado correctamente.")
    print("El cliente fue ubicado al final de la cola.")
def atender_siguiente_cliente(cola, pila):
    print("\n====================================")
    print("       ATENDER SIGUIENTE CLIENTE")
    print("====================================")
    if cola.esta_vacia():
        print("ERROR: no hay clientes en espera.")
        return
    cliente = cola.atender()
    pila.apilar(cliente)
    print("Cliente atendido:", cliente)
    print("La atencion fue registrada en la pila.")
def menu_cola(cola, pila):
    while True:
        print("\n====================================")
        print("       COLA DE CLIENTES")
        print("====================================")
        print("1. Registrar clientes iniciales")
        print("2. Agregar nuevo cliente")
        print("3. Atender siguiente cliente")
        print("4. Mostrar clientes en espera")
        print("5. Volver al menu principal")
        print("====================================")
        opcion = pedir_entero(
            "Seleccione una opcion: ",
            1
        )
        if opcion == 1:
            registrar_clientes_iniciales(cola)
        elif opcion == 2:
            agregar_nuevo_cliente(cola)
        elif opcion == 3:
            atender_siguiente_cliente(
                cola,
                pila
            )
        elif opcion == 4:
            cola.mostrar()
        elif opcion == 5:
            return
        else:
            print(
                "ERROR: opcion no valida. "
                "Seleccione entre 1 y 5."
            )
def mostrar_ultimo_atendido(pila):
    print("\n====================================")
    print("       ULTIMO CLIENTE ATENDIDO")
    print("====================================")
    cliente = pila.ultimo_atendido()
    if cliente is None:
        print(
            "ERROR: no hay clientes atendidos."
        )
    else:
        print(
            "Ultimo cliente atendido:",
            cliente
        )
def deshacer_ultima_atencion(cola, pila):
    print("\n====================================")
    print("       DESHACER ULTIMA ATENCION")
    print("====================================")
    if pila.esta_vacia():
        print(
            "ERROR: no hay ninguna atencion "
            "para deshacer."
        )
        return
    cliente = pila.desapilar()
    cola.insertar_al_frente(cliente)
    print(
        "Se deshizo la atencion de:",
        cliente
    )
    print(
        "El cliente regreso al inicio de la cola."
    )
def menu_pila(cola, pila):
    while True:
        print("\n====================================")
        print("       PILA DE OPERACIONES")
        print("====================================")
        print("1. Mostrar ultimo cliente atendido")
        print("2. Deshacer ultima atencion")
        print("3. Mostrar historial")
        print("4. Volver al menu principal")
        print("====================================")
        opcion = pedir_entero(
            "Seleccione una opcion: ",
            1
        )
        if opcion == 1:
            mostrar_ultimo_atendido(pila)
        elif opcion == 2:
            deshacer_ultima_atencion(
                cola,
                pila
            )
        elif opcion == 3:
            pila.mostrar()
        elif opcion == 4:
            return
        else:
            print(
                "ERROR: opcion no valida. "
                "Seleccione entre 1 y 4."
            )
def agregar_producto(lista):
    print("\n====================================")
    print("          AGREGAR PRODUCTO")
    print("====================================")
    while True:
        codigo = pedir_entero(
            "Ingrese el codigo del producto: ",
            1
        )
        producto = lista.buscar(codigo)
        if producto is not None:
            print(
                "ERROR: ya existe un producto "
                "con ese codigo."
            )
        else:
            break
    nombre = pedir_texto(
        "Ingrese el nombre del producto: "
    )
    precio = pedir_decimal(
        "Ingrese el precio del producto: ",
        0.01
    )
    lista.agregar(
        codigo,
        nombre,
        precio
    )
    print("\nProducto agregado correctamente.")
def buscar_producto(lista):
    print("\n====================================")
    print("          BUSCAR PRODUCTO")
    print("====================================")
    codigo = pedir_entero(
        "Ingrese el codigo del producto: ",
        1
    )
    producto = lista.buscar(codigo)
    if producto is None:
        print(
            "ERROR: no existe un producto "
            "con ese codigo."
        )
    else:
        print("\nPRODUCTO ENCONTRADO")
        print("------------------------------------")
        print("Codigo :", producto.codigo)
        print("Nombre :", producto.nombre)
        print(
            "Precio : $",
            format(producto.precio, ".2f")
        )
        print("------------------------------------")
def eliminar_producto(lista):
    print("\n====================================")
    print("          ELIMINAR PRODUCTO")
    print("====================================")
    codigo = pedir_entero(
        "Ingrese el codigo del producto: ",
        1
    )
    eliminado = lista.eliminar(codigo)
    if eliminado:
        print(
            "Producto eliminado correctamente."
        )
    else:

        print(
            "ERROR: no existe un producto "
            "con ese codigo."
        )
def menu_productos(lista):
    while True:
        print("\n====================================")
        print("       LISTA DE PRODUCTOS")
        print("====================================")
        print("1. Agregar producto")
        print("2. Mostrar de inicio a fin")
        print("3. Mostrar de fin a inicio")
        print("4. Buscar producto")
        print("5. Eliminar producto")
        print("6. Volver al menu principal")
        print("====================================")
        opcion = pedir_entero(
            "Seleccione una opcion: ",
            1
        )
        if opcion == 1:
            agregar_producto(lista)
        elif opcion == 2:
            lista.mostrar_inicio_fin()
        elif opcion == 3:
            lista.mostrar_fin_inicio()
        elif opcion == 4:
            buscar_producto(lista)
        elif opcion == 5:
            eliminar_producto(lista)
        elif opcion == 6:
            return
        else:
            print(
                "ERROR: opcion no valida. "
                "Seleccione entre 1 y 6."
            )
def main():
    cola = ColaClientes()
    pila = PilaAtendidos()
    lista_productos = ListaDobleProductos()
    while True:
        print("\n====================================")
        print("       SISTEMA DE ATENCION")
        print("====================================")
        print("1. Cola de clientes")
        print("2. Pila de operaciones")
        print("3. Lista de productos")
        print("4. Salir")
        print("====================================")
        opcion = pedir_entero(
            "Seleccione una opcion: ",
            1
        )
        if opcion == 1:
            menu_cola(
                cola,
                pila
            )
        elif opcion == 2:
            menu_pila(
                cola,
                pila
            )
        elif opcion == 3:
            menu_productos(
                lista_productos
            )
        elif opcion == 4:
            print("\n====================================")
            print("Sistema finalizado correctamente.")
            print("Gracias por utilizar el sistema.")
            print("====================================")
            break
        else:
            print(
                "ERROR: opcion no valida. "
                "Seleccione entre 1 y 4."
            )
main()
```

## Pruebas de Desempeño
Qué va aquí?

Medición empírica del tiempo de ejecución.

Gráfica de comportamiento.

Relación entre la gráfica obtenida y el análisis de complejidad teórico.


```python
import time
import matplotlib.pyplot as plt

def bench_cola(n: int, step: int = 500) -> list[tuple[int, float]]:
    print("Generando datos (Cola - Agregar N elementos) ", end='')
    results: list[tuple[int, float]] = []

    for size in range(1, n, step):
        cola = ColaClientes()
        
        ti = time.perf_counter()
        
        for i in range(size):
            cola.agregar(f"Cliente_{i}")
            
        tf = time.perf_counter()

        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
        
    print(" Listo!")
    return results

def bench_lista_buscar(n: int, step: int = 500) -> list[tuple[int, float]]:
    print("Generando datos (Lista Doble - Buscar peor caso) ", end='')
    results: list[tuple[int, float]] = []

    for size in range(1, n, step):
        lista = ListaDobleProductos()
        for i in range(size):
            lista.agregar(i, f"Prod_{i}", 1000)
        
        id_a_buscar = size - 1 

        ti = time.perf_counter()
        
        lista.buscar(id_a_buscar) 
        
        tf = time.perf_counter()

        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
        
    print(" Listo!")
    return results

def dibujar_resultados(resultados_cola, resultados_lista, dimensiones=(12, 5)):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=dimensiones)
    
    x_cola = [size for size, timing in resultados_cola]
    y_cola = [timing for size, timing in resultados_cola]
    ax1.plot(x_cola, y_cola, color='blue', marker='o')
    ax1.set_title("Cola: Tiempo de agregar N clientes")
    ax1.set_xlabel("Tamaño de entrada (N)")
    ax1.set_ylabel("Tiempo de ejecución (ms)")
    ax1.grid(True)

    x_lista = [size for size, timing in resultados_lista]
    y_lista = [timing for size, timing in resultados_lista]
    ax2.plot(x_lista, y_lista, color='red', marker='s')
    ax2.set_title("Lista Doble: Tiempo de Búsqueda (Peor Caso)")
    ax2.set_xlabel("Tamaño de la lista (N)")
    ax2.set_ylabel("Tiempo de ejecución (ms)")
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

print("Iniciando Pruebas de Desempeño...")
tiempos_cola = bench_cola(10000, step=1000)
tiempos_lista = bench_lista_buscar(10000, step=1000)

dibujar_resultados(tiempos_cola, tiempos_lista)
```
