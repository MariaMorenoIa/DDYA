## Ejercicio: Dar cambio con monedas colombianas
Una tienda en Colombia necesita desarrollar un programa que calcule la menor cantidad de monedas necesarias para entregar un cambio determinado a un cliente.

La tienda dispone de las siguientes monedas:

monedas = [50, 100, 200, 500, 1000]
El usuario deberá ingresar el valor del cambio que necesita entregar.

Por ejemplo:
Ingrese el valor del cambio: 1850
El programa debe determinar cuál es la menor cantidad de monedas necesaria para completar exactamente ese valor.

Una posible solución sería:
1000 + 500 + 200 + 100 + 50 = 1850

Por lo tanto:
Cantidad mínima de monedas: 5

## 8) Código funcional **[10 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def leer_cambio():
    while True:
        try:
            cambio = int(input("Ingrese el valor del cambio: "))
            if cambio > 0:
                return cambio
            else:
                print("Valor no válido. Debe ingresar un número entero mayor que 0.")
        except ValueError:
            print("Valor no válido. Ingrese un número entero.")

def cambio_minimo_memo(monedas, valor, memo, seleccion):
    if valor in memo:
        return memo[valor]
    if valor == 0:
        return 0
    min_monedas = float('inf')
    mejor_moneda = None
    for m in monedas:
        if valor - m >= 0:
            num_monedas = 1 + cambio_minimo_memo(monedas, valor - m, memo, seleccion)
            if num_monedas < min_monedas:
                min_monedas = num_monedas
                mejor_moneda = m
    memo[valor] = min_monedas
    seleccion[valor] = mejor_moneda
    return min_monedas

def reconstruir_solucion(seleccion, valor):
    resultado = []
    while valor > 0 and seleccion.get(valor) is not None:
        resultado.append(seleccion[valor])
        valor -= seleccion[valor]
    return resultado

def main():
    monedas = [50, 100, 200, 500, 1000]
    cambio = leer_cambio()

    # Ajuste: aproximar al múltiplo más cercano de la moneda más pequeña (50)
    if cambio % 50 != 0:
        aproximado = round(cambio / 50) * 50
        print(f"El valor {cambio} no puede representarse exactamente con las monedas disponibles.")
        print(f"Se aproximará al valor más cercano: {aproximado}")
        cambio = aproximado

    memo = {}
    seleccion = {}
    cantidad = cambio_minimo_memo(monedas, cambio, memo, seleccion)
    if cantidad == float('inf'):
        print(f"No es posible entregar un cambio exacto de {cambio} con las monedas disponibles.")
    else:
        combinacion = reconstruir_solucion(seleccion, cambio)
        print("Cantidad mínima de monedas:", cantidad)
        print("Monedas necesarias:", " + ".join(map(str, combinacion)), "=", cambio)

main()


```

```python
# Calcula el tiempo de ejecución con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecución
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA

# Asegúrate de tener instalada la librería:
# pip install matplotlib


import time
import matplotlib.pyplot as plt

def bench(n: int, step: int = 50) -> list[tuple[int, float]]:
    monedas = [50, 100, 200, 500, 1000]
    results: list[tuple[int, float]] = []
    for size in range(step, n + 1, step):
        memo = {}
        seleccion = {}
        ti = time.perf_counter()
        cambio_minimo_memo(monedas, size, memo, seleccion)
        tf = time.perf_counter()
        delta = (tf - ti) * 1000  # tiempo en milisegundos
        results.append((size, delta))
    return results

def dibujar(x, y, dimensiones=(7, 5)):
    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker="o", color="red")
    plt.xlabel("Valor del cambio")
    plt.ylabel("Tiempo de ejecución (ms)")
    plt.title("Desempeño del algoritmo de cambio mínimo")
    plt.grid(True)
    plt.show()


if __name__ == "__main__":
    main()
    # Pruebas de desempeño
    tiempos = bench(2000, step=100)
    x = [size for size, timing in tiempos]
    y = [timing for size, timing in tiempos]
    dibujar(x, y)
```
