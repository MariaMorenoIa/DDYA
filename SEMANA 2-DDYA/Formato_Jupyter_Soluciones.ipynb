Punto 2. Orden descendente

# Codigo

```python
def registrar_producto(n):
    productos = []
    for i in range(n):
        precio = float(input("Ingrese el precio del producto: "))
        productos.append(precio)
    return productos
def ord_ascen(productos):
    for j in range(1, len(productos)):
        key = productos[j]
        i = j - 1
        while i >= 0 and productos[i] > key:
            productos[i + 1] = productos[i]
            i = i - 1
        productos[i + 1] = key
    return productos
def ord_descen(productos):
    for j in range(1, len(productos)):
        key = productos[j]
        i = j - 1

        while i >= 0 and productos[i] < key:
            productos[i + 1] = productos[i]
            i = i - 1

        productos[i + 1] = key
    return productos
def main():
    n = int(input("Ingrese la cantidad de productos: "))
    productos = registrar_producto(n)
    tipo = input("Ingrese el tipo de ordenamiento (ascendente/descendente): ")
    if tipo.lower() == "ascendente":
        productos = ord_ascen(productos)
    elif tipo.lower() == "descendente":
        productos = ord_descen(productos)
    else:
        print("Tipo de ordenamiento no válido")
        return
    print("Precios ordenados:")
    for precio in productos:
        print(precio)
main()
```
