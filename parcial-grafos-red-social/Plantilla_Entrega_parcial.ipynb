### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación
- Función T[n]
- Mejor y peor caso con gráfica


**Justificación:** El sistema utiliza un grafo para representar una red de usuarios, donde los vértices (V) representan a los usuarios y las aristas (E) representan las amistades entre ellos. Para analizar la complejidad, se tiene en cuenta el número de vértices y aristas del grafo.

La construcción del grafo y la impresión de su lista de adyacencia tienen una complejidad de O(V + E). Los algoritmos BFS (búsqueda en anchura) y DFS (búsqueda en profundidad) visitan los vértices y examinan las aristas para recorrer la red, por lo que su complejidad temporal en el peor caso es O(V + E).

Cuando se utiliza BFS para encontrar la ruta más corta en un grafo no ponderado, el recorrido tiene una complejidad de O(V + E), mientras que la reconstrucción de la ruta tiene una complejidad máxima de O(V). Por lo tanto, la complejidad total continúa siendo O(V + E).

El espacio auxiliar utilizado por BFS y DFS es O(V), debido a las estructuras necesarias para registrar los vértices visitados, la cola o pila y, cuando corresponde, las distancias y los predecesores.

**Función \(T(n)\)**

La función de complejidad temporal se expresa como:
\
T(V,E)=O(V+E)
\

Donde:
- \(V\): número de vértices o usuarios.
- \(E\): número de aristas o amistades.
- \(O(V+E)\): representa el crecimiento del trabajo computacional según el tamaño del grafo.
Para simplificar la gráfica, se puede considerar una red en la que el número de aristas crece proporcionalmente al número de vértices. En ese caso, el crecimiento ilustrativo es lineal, es decir, \(O(V)\).

La siguiente gráfica compara el crecimiento ilustrativo del mejor y el peor caso. Se utiliza una función constante para representar el mejor caso y una función lineal para representar el peor caso, suponiendo que el número de aristas crece proporcionalmente al número de vértices. La gráfica es teórica y no representa mediciones de tiempo real.

```python

import matplotlib.pyplot as plt
vertices = [5, 10, 20, 30, 40, 50]

mejor_caso = [1 for v in vertices]

peor_caso = [v + 2 * v for v in vertices]

plt.figure(figsize=(8, 5))

plt.plot(
    vertices, mejor_caso,
    label="Mejor caso: O(1)",
    marker="o"
)

plt.plot(
    vertices, peor_caso,
    label="Peor caso: O(V + E)",
    marker="o"
)

plt.title("Comparación de mejor y peor caso en grafos")
plt.xlabel("Número de vértices (V)")
plt.ylabel("Número relativo de operaciones")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

la gráfica compara casos ilustrativos de una búsqueda que puede terminar inmediatamente y otra que recorre toda la red. No significa que BFS y DFS tengan complejidad constante en general: su complejidad en el peor caso es \(O(V+E)\). Además, si la búsqueda debe recorrer todos los vértices, incluso el mejor caso puede dejar de ser \(O(1)\).

## 6) Código funcional **[15 pts]**

```python
from collections import deque

def crear_grafo_inicial():
    grafo = {
        "Ana": [],
        "Carlos": [],
        "Laura": [],
        "Mateo": [],
        "Sofía": [],
        "Daniel": [],
        "Valentina": []
    }
    amistades = [
        ("Ana", "Carlos"),
        ("Ana", "Laura"),
        ("Carlos", "Mateo"),
        ("Carlos", "Sofía"),
        ("Laura", "Daniel"),
        ("Sofía", "Valentina"),
        ("Daniel", "Valentina")
    ]
    for persona1, persona2 in amistades:
        agregar_amistad(grafo, persona1, persona2)
    return grafo


def normalizar_nombre(nombre):
    return nombre.strip()


def agregar_amistad(grafo, persona1, persona2):
    """Agrega una amistad no dirigida, sin duplicados."""
    persona1 = normalizar_nombre(persona1)
    persona2 = normalizar_nombre(persona2)
    if persona1 not in grafo or persona2 not in grafo:
        raise ValueError("Ambas personas deben existir en el grafo.")
    if persona1 == persona2:
        raise ValueError("Una persona no puede conectarse consigo misma.")
    if persona2 not in grafo[persona1]:
        grafo[persona1].append(persona2)
    if persona1 not in grafo[persona2]:
        grafo[persona2].append(persona1)


def mostrar_grafo(grafo):
    print("Lista de adyacencia:")
    for usuario, vecinos in grafo.items():
        contactos = ", ".join(vecinos) if vecinos else "Sin contactos"
        print(f"{usuario}: {contactos}")


def validar_usuario(grafo, nombre):
    return normalizar_nombre(nombre) in grafo


def solicitar_usuario_existente(grafo, mensaje):
    """Solicita un nombre hasta que corresponda a un vértice existente."""
    while True:
        nombre = normalizar_nombre(input(mensaje))
        if nombre in grafo:
            return nombre
        print("Error: ese usuario no existe. Usuarios disponibles: " + ", ".join(grafo.keys()))


def bfs(grafo, origen):
    """Devuelve el orden BFS y las distancias mínimas desde el origen."""
    if origen not in grafo:
        raise ValueError(f"El usuario '{origen}' no existe.")
    cola = deque([origen])
    visitados = {origen}
    distancias = {origen: 0}
    orden = []
    while cola:
        actual = cola.popleft()
        orden.append(actual)
        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                distancias[vecino] = distancias[actual] + 1
                cola.append(vecino)
    return orden, distancias


def dfs(grafo, origen):
    """Devuelve el orden DFS usando recursividad."""
    if origen not in grafo:
        raise ValueError(f"El usuario '{origen}' no existe.")
    visitados = set()
    orden = []

    def visitar(actual):
        visitados.add(actual)
        orden.append(actual)
        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitar(vecino)

    visitar(origen)
    return orden


def ruta_minima(grafo, origen, destino):
    """Retorna una ruta de menor número de aristas usando BFS, o None."""
    if origen not in grafo or destino not in grafo:
        raise ValueError("El origen y el destino deben existir en el grafo.")
    cola = deque([origen])
    visitados = {origen}
    predecesor = {origen: None}

    while cola:
        actual = cola.popleft()
        if actual == destino:
            break
        for vecino in grafo[actual]:
            if vecino not in visitados:
                visitados.add(vecino)
                predecesor[vecino] = actual
                cola.append(vecino)

    if destino not in predecesor:
        return None

    ruta = []
    actual = destino
    while actual is not None:
        ruta.append(actual)
        actual = predecesor[actual]
    ruta.reverse()
    return ruta


def agregar_usuario_interactivo(grafo):
    while True:
        nuevo = normalizar_nombre(input("Nombre del nuevo usuario: "))
        if not nuevo:
            print("Error: el nombre no puede estar vacío.")
        elif nuevo in grafo:
            print("Error: ese usuario ya existe. Ingresa otro nombre.")
        else:
            break

    conexiones = []
    print("Ingresa al menos dos usuarios existentes para conectar con el nuevo usuario.")
    while len(conexiones) < 2:
        nombre = normalizar_nombre(input(f"Conexión {len(conexiones) + 1}: "))
        if nombre not in grafo:
            print("Error: ese usuario no existe. Usuarios disponibles: " + ", ".join(grafo.keys()))
        elif nombre == nuevo:
            print("Error: el nuevo usuario no puede conectarse consigo mismo.")
        elif nombre in conexiones:
            print("Error: esa conexión ya fue ingresada. Elige otra persona.")
        else:
            conexiones.append(nombre)

    grafo[nuevo] = []
    for persona in conexiones:
        agregar_amistad(grafo, nuevo, persona)
    print(f"Usuario {nuevo} agregado con {len(conexiones)} conexiones.")
    mostrar_grafo(grafo)


def menu():
    grafo = crear_grafo_inicial()
    while True:
        print("\nMENÚ PRINCIPAL")
        print("1. Mostrar red social")
        print("2. Recorrer con BFS")
        print("3. Recorrer con DFS")
        print("4. Encontrar ruta mínima entre dos usuarios")
        print("5. Agregar un usuario")
        print("6. Salir")
        opcion = input("Selecciona una opción: ").strip()

        if opcion == "1":
            mostrar_grafo(grafo)
        elif opcion == "2":
            origen = solicitar_usuario_existente(grafo, "Usuario inicial para BFS: ")
            orden, distancias = bfs(grafo, origen)
            print("Orden BFS: " + " -> ".join(orden))
            print("Grados de separación:")
            for grado in sorted(set(distancias.values())):
                personas = [u for u in orden if distancias[u] == grado]
                print(f"Grado {grado}: " + ", ".join(personas))
            no_alcanzables = [u for u in grafo if u not in distancias]
            if no_alcanzables:
                print("Sin conexión con el origen: " + ", ".join(no_alcanzables))
        elif opcion == "3":
            origen = solicitar_usuario_existente(grafo, "Usuario inicial para DFS: ")
            print("Orden DFS: " + " -> ".join(dfs(grafo, origen)))
        elif opcion == "4":
            origen = solicitar_usuario_existente(grafo, "Usuario inicial: ")
            destino = solicitar_usuario_existente(grafo, "Usuario destino: ")
            ruta = ruta_minima(grafo, origen, destino)
            if ruta is None:
                print(f"No existe una conexión entre {origen} y {destino}.")
            else:
                print("Ruta encontrada: " + " -> ".join(ruta))
                print("Grados de separación (conexiones): " + str(len(ruta) - 1))
        elif opcion == "5":
            agregar_usuario_interactivo(grafo)
        elif opcion == "6":
            print("Programa finalizado.")
            break
        else:
            print("Error: opción no válida. Selecciona una opción del menú.")
menu()
```

```python

grafo_prueba = crear_grafo_inicial()

# 1. Lista de adyacencia y amistades en ambos sentidos
assert "Carlos" in grafo_prueba["Ana"]
assert "Ana" in grafo_prueba["Carlos"]
assert len(grafo_prueba["Ana"]) == 2

# 2. BFS y grados de separación desde Ana
orden_bfs, distancias = bfs(grafo_prueba, "Ana")
assert orden_bfs[0] == "Ana"
assert distancias["Ana"] == 0
assert distancias["Carlos"] == 1
assert distancias["Laura"] == 1
assert distancias["Valentina"] == 3

# 3. DFS visita cada usuario alcanzable una sola vez
orden_dfs = dfs(grafo_prueba, "Ana")
assert len(orden_dfs) == len(set(orden_dfs)) == len(grafo_prueba)

# 4. Ruta mínima entre Ana y Valentina: 3 conexiones
ruta = ruta_minima(grafo_prueba, "Ana", "Valentina")
assert ruta is not None
assert ruta[0] == "Ana" and ruta[-1] == "Valentina"
assert len(ruta) - 1 == 3

# 5. Ruta de una persona a sí misma
assert ruta_minima(grafo_prueba, "Ana", "Ana") == ["Ana"]

# 6. Agregar amistad no duplica vecinos
cantidad = len(grafo_prueba["Ana"])
agregar_amistad(grafo_prueba, "Ana", "Carlos")
assert len(grafo_prueba["Ana"]) == cantidad

# 7. Los nombres inexistentes generan error en los algoritmos
try:
    bfs(grafo_prueba, "UsuarioInventado")
    raise AssertionError("BFS debió rechazar un usuario inexistente.")
except ValueError:
    pass

try:
    ruta_minima(grafo_prueba, "Ana", "UsuarioInventado")
    raise AssertionError("La ruta debió rechazar un destino inexistente.")
except ValueError:
    pass

# 8. El nuevo usuario queda conectado en ambos sentidos
grafo_prueba["Juliana"] = []
agregar_amistad(grafo_prueba, "Juliana", "Mateo")
agregar_amistad(grafo_prueba, "Juliana", "Daniel")
assert "Juliana" in grafo_prueba["Mateo"]
assert "Juliana" in grafo_prueba["Daniel"]
assert len(grafo_prueba["Juliana"]) == 2

print("Todas las pruebas se ejecutaron correctamente.")
print("BFS desde Ana:", " -> ".join(orden_bfs))
print("Distancias desde Ana:", distancias)
print("DFS desde Ana:", " -> ".join(orden_dfs))
print("Ruta mínima Ana-Valentina:", " -> ".join(ruta))
mostrar_grafo(grafo_prueba)

```
