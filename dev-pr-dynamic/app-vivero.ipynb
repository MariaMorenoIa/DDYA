## 1) Descripción del problema **[1 pts]**


Un vivero está analizando el crecimiento de la producción de nuevas plantas a partir de un 
proceso de reproducción controlada.

Se ha observado que el número de nuevas plantas producidas en el mes n sigue el siguiente 
comportamiento:

• En el mes 0 se produjeron 0 nuevas plantas.

• En el mes 1 se produjo 1 nueva planta.

• A partir del mes 2, el número de nuevas plantas producidas es igual a la suma de las plantas 

producidas durante los dos meses anteriores, debido a que las plantas que alcanzan su etapa de 
reproducción generan nuevos brotes siguiendo este patrón de crecimiento.

Relación de recurrencia:

T(n) = 0 si n = 0

T(n) = 1 si n = 1

T(n) = T(n-1) + T(n-2) si n ≥ 2

Determinar con memoización cuántas nuevas plantas serán producidas en el mes n.

*Conclusión*: Aunque la relación de recurrencia T(n) = T(n-1) + T(n-2) corresponde a Fibonacci y tendría un crecimiento exponencial si se implementara directamente, la memoización evita los cálculos repetidos, por lo que el algoritmo implementado tiene una complejidad temporal de Θ(n) en el peor caso y espacial de Θ(n).

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python

memo = {}
def calcular_plantas(n):
    if n == 0:
        memo[0] = 0
        return memo[0]
    if n == 1:
        memo[1] = 1
        return memo[1]
    if n in memo:
        return memo[n]
    memo[n] = calcular_plantas(n - 1) + calcular_plantas(n - 2)
    return memo[n]
def main():
    print("   PRODUCCIÓN DE NUEVAS PLANTAS")
    print("==========================================")
    while True:
        try:
            entrada = input("Ingrese el valor del mes (n): ")
            n = int(entrada)
            if n < 0:
                print("Error de entrada: el mes debe ser un entero mayor o igual a 0.")
                print("Por favor, inténtelo nuevamente.\n")
                continue
            break
        except ValueError:
            print("Error de entrada: debe ingresar un número entero.")
            print("Por favor, inténtelo nuevamente.\n")
    if n in memo:
        resultado = memo[n]
    else:
        resultado = calcular_plantas(n)
    print("\n**************************************")
    print("Resultado:")
    print(f"En el mes {n} se producirán {resultado} nuevas plantas.")
    print("**************************************")


if __name__ == "__main__":
    main()
```
