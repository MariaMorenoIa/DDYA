## Ejercicio: Organización y búsqueda de productos en una tienda
Una tienda maneja una lista con los códigos numéricos de sus productos. Como los productos se registran a medida que llegan, los códigos inicialmente se encuentran desordenados.

El encargado necesita desarrollar un programa que permita:


Registrar una cantidad n de productos.
Ordenar sus códigos utilizando Merge Sort.
Buscar un código específico utilizando Búsqueda Binaria.

Ejemplo

El usuario indica que desea registrar 6 productos:

Ingrese la cantidad de productos: 6 
Ingrese el código del producto 1: 105
 Ingrese el código del producto 2: 32
  Ingrese el código del producto 3: 78 
  Ingrese el código del producto 4: 15 
  Ingrese el código del producto 5: 120 
  Ingrese el código del producto 6: 50

La lista original sería:

[105, 32, 78, 15, 120, 50]

*Punto 1. Ordenar los productos con Merge Sort*

Utilizando el algoritmo Merge Sort, organice los códigos de los productos de menor a mayor.

Resultado esperado:

[15, 32, 50, 78, 105, 120]
Recuerde que Merge Sort utiliza la estrategia de Dividir y Conquistar:

Dividir: separar la lista en dos partes.
Conquistar: ordenar cada mitad.
Combinar: unir nuevamente las mitades de forma ordenada.

*Punto 2. Buscar un producto con Búsqueda Binaria*

Después de ordenar la lista, el usuario deberá ingresar el código del producto que desea buscar.

Ejemplo:

Ingrese el código que desea buscar: 78
El programa deberá utilizar Búsqueda Binaria para determinar si el producto existe.

Salida:

El producto con código 78 fue encontrado.
Si el usuario busca:

Ingrese el código que desea buscar: 90
El programa deberá mostrar:

El producto con código 90 no se encuentra registrado.

```python
def registra_cant_prod():
    while True:
        try:
            n = int(input("Ingrese la cantidad de productos que desea registrar: "))
            if n > 0:
                return n
            else:
                print("Debe ingresar un número entero mayor que 0.")
        except ValueError:
            print("Entrada inválida. Ingrese un número entero.")

def ord_mergesort(A):
    if len(A) > 1:
        mid = len(A) // 2
        L = A[:mid]
        R = A[mid:]
        ord_mergesort(L)
        ord_mergesort(R)
        i = j = k = 0
        while i < len(L) and j < len(R):
            if L[i] <= R[j]:
                A[k] = L[i]
                i += 1
            else:
                A[k] = R[j]
                j += 1
            k += 1
        while i < len(L):
            A[k] = L[i]
            i += 1
            k += 1
        while j < len(R):
            A[k] = R[j]
            j += 1
            k += 1
    return A

def busc_binaria(A, x):
    lo, hi = 0, len(A) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if A[mid] == x:
            return mid
        elif A[mid] < x:
            lo = mid + 1
        else:
            hi = mid - 1
    return None

def main():
    cantprod = registra_cant_prod()
    listaprod = []
    for i in range(cantprod):
        while True:
            try:
                codigo = int(input(f"Ingrese el código del producto {i+1}: "))
                listaprod.append(codigo)
                break
            except ValueError:
                print("Entrada inválida. Ingrese un número entero.")

    print("La lista original es:", listaprod)

    listaord = ord_mergesort(listaprod)
    print("La lista organizada es:", listaord)

    while True:
        try:
            buscado = int(input("Ingrese el código que desea buscar: "))
            break
        except ValueError:
            print("Entrada inválida. Ingrese un número entero.")

    pos = busc_binaria(listaord, buscado)
    if pos is not None:
        print(f"El producto con código {buscado} fue encontrado en la posición {pos+1}.")
    else:
        print(f"El producto con código {buscado} no se encuentra registrado.")

main()

```
