## 1) Descripción del problema **[1 pts]**


Una tienda necesita organizar los productos de su inventario para consultar rápidamente la información de cada uno. Cada producto tiene un código numérico único, un nombre, un precio y una cantidad disponible.

Cuando la tienda tiene muchos productos, buscar uno recorriendo una lista completa puede ser poco eficiente. Para mejorar este proceso se implementa una Tabla Hash, utilizando el código numérico como llave y la función:
\[
h(código) = código \bmod m
\]
donde m es el tamaño de la tabla. Si varios códigos producen el mismo índice, se presenta una colisión. Para resolverla se utiliza encadenamiento, es decir, cada posición de la tabla almacena una lista de productos.

El programa permite registrar productos, buscar por código, eliminar un producto, actualizar su cantidad y mostrar el contenido completo de la tabla. La estructura se implementa manualmente con listas de Python, sin utilizar diccionarios como estructura principal de almacenamiento.

**Justificación**

La función hash calcula directamente la posición del producto mediante el residuo entre el código y el tamaño de la tabla. Esta operación tiene complejidad constante O(1).


Con encadenamiento, cada índice contiene una lista de productos. Si la función hash distribuye los códigos uniformemente, cada cadena tendrá pocos elementos y las operaciones de búsqueda, inserción y eliminación tienen un costo promedio O(1).


En el peor caso, todos los productos pueden quedar en la misma posición. En ese escenario, una búsqueda o eliminación debe revisar hasta los n productos de la cadena, por lo que su complejidad es O(n).

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

```python

def es_entero(texto):
   
    texto = texto.strip()
    if not texto:
        return False
    if texto[0] in "+-":
        return len(texto) > 1 and texto[1:].isdigit()
    return texto.isdigit()


def solicitar_entero(mensaje, minimo=None):
    
    while True:
        valor = input(mensaje).strip()
        if not es_entero(valor):
            print("Error: debe ingresar un numero entero.")
            continue
        numero = int(valor)
        if minimo is not None and numero < minimo:
            print(f"Error: el valor debe ser mayor o igual que {minimo}.")
            continue
        return numero


def solicitar_numero(mensaje, minimo=None):
    
    while True:
        valor = input(mensaje).strip().replace(",", ".")
        try:
            numero = float(valor)
        except ValueError:
            print("Error: debe ingresar un valor numerico.")
            continue
        if minimo is not None and numero < minimo:
            print(f"Error: el valor debe ser mayor o igual que {minimo}.")
            continue
        return numero


def solicitar_texto(mensaje):
    
    while True:
        valor = input(mensaje).strip()
        if valor:
            return valor
        print("Error: este campo no puede estar vacio.")


def crear_producto(codigo, nombre, precio, cantidad):
    
    return {
        "codigo": codigo,
        "nombre": nombre,
        "precio": precio,
        "cantidad": cantidad
    }


class TablaHash:
    

    def __init__(self, m):
        if not isinstance(m, int) or m <= 0:
            raise ValueError("El tamano de la tabla debe ser un entero positivo.")
        self.m = m
        self.tabla = [[] for _ in range(m)]
        self.total_productos = 0

    def funcion_hash(self, codigo):
       
        return codigo % self.m

    def buscar(self, codigo):
       
        indice = self.funcion_hash(codigo)
        cadena = self.tabla[indice]

        for posicion, producto in enumerate(cadena):
            if producto["codigo"] == codigo:
                return producto, indice, posicion

        return None, indice, -1

    def insertar(self, producto):
       
        codigo = producto["codigo"]
        existente, indice, _ = self.buscar(codigo)

        if existente is not None:
            return False, indice, True

        hubo_colision = len(self.tabla[indice]) > 0
        self.tabla[indice].append(producto)
        self.total_productos += 1
        return True, indice, hubo_colision

    def eliminar(self, codigo):
       
        producto, indice, posicion = self.buscar(codigo)

        if producto is None:
            return None, indice

        eliminado = self.tabla[indice].pop(posicion)
        self.total_productos -= 1
        return eliminado, indice

    def actualizar_cantidad(self, codigo, nueva_cantidad):
       
        producto, indice, _ = self.buscar(codigo)

        if producto is None:
            return None, indice

        producto["cantidad"] = nueva_cantidad
        return producto, indice

    def mostrar(self):
        
        print("\n" + "=" * 72)
        print("ESTADO COMPLETO DE LA TABLA HASH")
        print("=" * 72)

        for indice, cadena in enumerate(self.tabla):
            if not cadena:
                print(f"Indice {indice}: Vacio")
            else:
                print(f"Indice {indice}:")
                for producto in cadena:
                    print(
                        f"  {producto['codigo']} - {producto['nombre']} | "
                        f"Precio: ${producto['precio']:,.2f} | "
                        f"Cantidad: {producto['cantidad']}"
                    )

        print("-" * 72)
        print(f"Total de productos registrados: {self.total_productos}")
        print(f"Tamano de la tabla: {self.m}")
        print("=" * 72)

    def registrar_varios_interactivo(self):
        
        n = solicitar_entero("Cuantos productos desea registrar: ", minimo=1)

        for i in range(n):
            print(f"\nRegistro del producto {i + 1} de {n}")

            while True:
                codigo = solicitar_entero("Codigo numerico: ", minimo=0)
                existente, indice, _ = self.buscar(codigo)
                if existente is not None:
                    print("Error: ya existe un producto con ese codigo.")
                else:
                    break

            nombre = solicitar_texto("Nombre: ")
            precio = solicitar_numero("Precio: $", minimo=0)
            cantidad = solicitar_entero("Cantidad disponible: ", minimo=0)

            producto = crear_producto(codigo, nombre, precio, cantidad)
            insertado, indice, hubo_colision = self.insertar(producto)

            if insertado:
                print(f"Producto registrado en el indice {indice}.")
                if hubo_colision:
                    print("Se presento una colision y se resolvio por encadenamiento.")

    def buscar_interactivo(self):
        codigo = solicitar_entero("Ingrese el codigo del producto: ", minimo=0)
        producto, indice, posicion = self.buscar(codigo)

        print(f"Indice calculado: {indice}")
        if producto is None:
            print("El producto no esta registrado.")
        else:
            print("Producto encontrado:")
            mostrar_producto(producto)
            print(f"Posicion dentro de la cadena: {posicion}")

    def eliminar_interactivo(self):
        codigo = solicitar_entero("Codigo del producto que desea eliminar: ", minimo=0)
        producto, indice = self.eliminar(codigo)

        if producto is None:
            print("No se encontro el producto.")
        else:
            print(f"Producto eliminado del indice {indice}:")
            mostrar_producto(producto)

    def actualizar_interactivo(self):
        codigo = solicitar_entero("Codigo del producto que desea actualizar: ", minimo=0)
        producto, indice, _ = self.buscar(codigo)

        if producto is None:
            print("El producto no esta registrado.")
            return

        print("Producto encontrado:")
        mostrar_producto(producto)
        nueva_cantidad = solicitar_entero("Ingrese la nueva cantidad: ", minimo=0)
        producto_actualizado, _ = self.actualizar_cantidad(codigo, nueva_cantidad)

        print("Inventario actualizado correctamente.")
        mostrar_producto(producto_actualizado)


def mostrar_producto(producto):
    
    if producto is None:
        print("No hay informacion para mostrar.")
        return

    print(f"Codigo: {producto['codigo']}")
    print(f"Nombre: {producto['nombre']}")
    print(f"Precio: ${producto['precio']:,.2f}")
    print(f"Cantidad disponible: {producto['cantidad']}")


def cargar_datos_ejemplo(tabla):
    
    ejemplos = [
        crear_producto(101, "Arroz", 4500, 20),
        crear_producto(108, "Huevos", 16000, 12),
        crear_producto(115, "Azucar", 3800, 15),
        crear_producto(302, "Galletas", 5200, 18),
        crear_producto(312, "Cafe", 12000, 10),
        crear_producto(209, "Leche", 3800, 25),
        crear_producto(423, "Pan", 2500, 30)
    ]

    cargados = 0
    for producto in ejemplos:
        insertado, indice, colision = tabla.insertar(producto)
        if insertado:
            cargados += 1

    print(f"Se cargaron {cargados} productos de ejemplo.")
    print("Con tamano 7, los codigos 101, 108 y 115 colisionan en el indice 3.")
    tabla.mostrar()


def mostrar_menu():
    print("\n" + "=" * 54)
    print("       SISTEMA DE INVENTARIO - TABLA HASH")
    print("=" * 54)
    print("1. Registrar varios productos")
    print("2. Buscar un producto")
    print("3. Eliminar un producto")
    print("4. Actualizar cantidad disponible")
    print("5. Mostrar Tabla Hash completa")
    print("6. Cargar productos de ejemplo")
    print("7. Salir")
    print("=" * 54)


def ejecutar_programa():
    print("SISTEMA DE INVENTARIO UTILIZANDO TABLA HASH")
    print("Resolucion de colisiones mediante encadenamiento")

    m = solicitar_entero("Ingrese el tamano de la tabla hash (m): ", minimo=1)
    inventario = TablaHash(m)
    print(f"Tabla Hash creada correctamente con {m} posiciones vacias.")

    while True:
        mostrar_menu()
        opcion = solicitar_entero("Seleccione una opcion: ", minimo=1)

        if opcion == 1:
            inventario.registrar_varios_interactivo()
        elif opcion == 2:
            inventario.buscar_interactivo()
        elif opcion == 3:
            inventario.eliminar_interactivo()
        elif opcion == 4:
            inventario.actualizar_interactivo()
        elif opcion == 5:
            inventario.mostrar()
        elif opcion == 6:
            cargar_datos_ejemplo(inventario)
        elif opcion == 7:
            print("Programa finalizado. Gracias por utilizar el sistema.")
            break
        else:
            print("Opcion no valida. Seleccione una opcion entre 1 y 7.")



ejecutar_programa()

```

```python


def ejecutar_pruebas():
    tabla = TablaHash(7)

    # Creacion de la tabla y validacion de posiciones
    assert tabla.m == 7
    assert len(tabla.tabla) == 7
    assert tabla.total_productos == 0

    # Funcion hash y colisiones: 101, 108 y 115 generan indice 3
    assert tabla.funcion_hash(101) == 3
    assert tabla.funcion_hash(108) == 3
    assert tabla.funcion_hash(115) == 3

    p1 = crear_producto(101, "Arroz", 4500, 20)
    p2 = crear_producto(108, "Huevos", 16000, 12)
    p3 = crear_producto(115, "Azucar", 3800, 15)

    assert tabla.insertar(p1)[0] is True
    assert tabla.insertar(p2)[0] is True
    assert tabla.insertar(p3)[0] is True
    assert len(tabla.tabla[3]) == 3
    assert tabla.total_productos == 3

    # No permitir codigos duplicados
    duplicado = crear_producto(101, "Otro arroz", 5000, 5)
    assert tabla.insertar(duplicado)[0] is False
    assert tabla.total_productos == 3

    # Busqueda exitosa
    encontrado, indice, posicion = tabla.buscar(108)
    assert encontrado is not None
    assert encontrado["nombre"] == "Huevos"
    assert indice == 3 and posicion >= 0

    # Busqueda fallida
    no_encontrado, indice, posicion = tabla.buscar(999)
    assert no_encontrado is None and posicion == -1

    # Actualizacion de inventario
    actualizado, indice = tabla.actualizar_cantidad(101, 35)
    assert actualizado is not None
    assert actualizado["cantidad"] == 35

    # Eliminacion en una cadena con colisiones
    eliminado, indice = tabla.eliminar(108)
    assert eliminado is not None and eliminado["nombre"] == "Huevos"
    assert len(tabla.tabla[3]) == 2
    assert tabla.buscar(101)[0] is not None
    assert tabla.buscar(115)[0] is not None
    assert tabla.buscar(108)[0] is None

    # Eliminacion de codigo inexistente
    eliminado, indice = tabla.eliminar(777)
    assert eliminado is None

    # Consistencia del contador
    assert tabla.total_productos == 2
    total_contado = sum(len(cadena) for cadena in tabla.tabla)
    assert total_contado == tabla.total_productos

    print("Todas las pruebas finalizaron correctamente.")
    print("Se verifico creacion, hash, insercion, colision, duplicados,")
    print("busqueda exitosa, busqueda fallida, eliminacion y actualizacion.")
    print("Estado final utilizado en las pruebas:")
    tabla.mostrar()


ejecutar_pruebas()
```
